```python
import os
from dotenv import load_dotenv
from anthropic import Anthropic
```

```python
load_dotenv()
```

```python

client = Anthropic(
    base_url=os.getenv("BASE_URL"),
    api_key=os.getenv("API_KEY")
)
```

```python
message = client.messages.create(
    max_tokens=1024,                      # required by Anthropic
    messages=[
        {"role": "user", "content": "Suggest one name for a coffee shop. Reply with the name only."}
    ],
    model=os.getenv("MODEL_NAME"),
    extra_body={"temperature": 0.1}
)

# Extract and print response
print(message.content[0].text)
```

```python
import os 
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic(
    base_url=os.getenv("BASE_URL"),
    api_key=os.getenv("API_KEY"), 
)


message = client.messages.create(
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": "Suggest one name for a coffee shop. Reply with the name only.",
        }
    ],
    model=os.getenv("MODEL_NAME"),
    extra_body={"temperature": 1.5}
)

print(message.content[0].text)
```
